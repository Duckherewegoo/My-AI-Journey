```python
import os
from langchain_openai import ChatOpenAI
from langchain.agents import create_openai_tools_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import tool
from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse

# ===================== 1. 基础配置 =====================
# 替换为你的阿里云百炼 API Key
API_KEY = os.environ["DASHSCOPE_API_KEY"] 

# 定义工具（示例：必须有工具才能创建 agent）
@tool
def add_numbers(a: int, b: int) -> int:
    """计算两个数字的和"""
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """计算两个数字的积"""
    return a * b

tools = [add_numbers, multiply_numbers]

# ===================== 2. 正确实例化模型 =====================
# 定义不同模型（基础版/进阶版），model 参数传字符串模型名
def get_chat_model(model_name: str) -> ChatOpenAI:
    """封装模型实例化逻辑，避免重复代码"""
    return ChatOpenAI(
        api_key=API_KEY,
        base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
        model_name=model_name,  # 核心：传字符串模型名
        temperature=0.7
    )

# 基础模型（轻量版）和高级模型（旗舰版）
basic_model = get_chat_model("qwen-turbo")    # 基础版，适合短对话
advanced_model = get_chat_model("qwen-plus")  # 进阶版，适合长对话/复杂任务

# ===================== 3. 动态模型选择中间件 =====================
@wrap_model_call
def dynamic_model_selection(request: ModelRequest, handler) -> ModelResponse:
    """根据对话消息数量动态选择模型：消息>10条用高级模型，否则用基础模型"""
    # 获取对话历史消息数量
    message_count = len(request.state.get("messages", []))
    print(f"当前对话消息数：{message_count} → 选择模型：{'高级模型(qwen-plus)' if message_count>10 else '基础模型(qwen-turbo)'}")
    
    # 动态替换模型
    if message_count > 4:
        request.model = advanced_model
    else:
        request.model = basic_model
    
    # 执行原处理逻辑
    return handler(request)

# ===================== 4. 创建 Agent 并运行 =====================
# 定义 Agent 提示词模板（新版 LangChain 标准写法）
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个乐于助人的助手，使用提供的工具完成计算任务。"),
    ("user", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),  # 必须：存放思考过程
])

# 创建 agent（替代旧版 create_agent）
agent = create_openai_tools_agent(
    llm=basic_model,  # 默认模型
    tools=tools,
    prompt=prompt
)

# 创建 Agent 执行器，添加中间件
agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    middleware=[dynamic_model_selection],  # 注入动态模型选择中间件
    verbose=True,  # 打印详细执行过程
    handle_parsing_errors=True  # 容错：处理解析错误
)

# ===================== 5. 测试运行 =====================
# 测试1：短对话（使用基础模型）
print("\n===== 测试1：短对话 =====")
result1 = agent_executor.invoke({"input": "计算 100 + 2000 的结果"})
print("最终回答：", result1["output"])

# 测试2：模拟长对话（手动添加历史消息，触发高级模型）
print("\n===== 测试2：长对话 =====")
# 构造包含11条消息的历史（触发高级模型）
long_messages = [{"role": "user", "content": f"测试消息{i}"} for i in range(5)]
result2 = agent_executor.invoke({
    "input": "计算 3120 × 46.101 的结果",
    "chat_history": long_messages  # 传入历史消息，让 message_count > 4
})
print("最终回答：", result2["output"])
```

```python
import tiktoken

# 使用cl100k_base编码
text = "元宝。我现在好困，这句话如果用token分词的话该怎么分？"
encoder = tiktoken.get_encoding("cl100k_base")

# 编码
tokens = encoder.encode(text)
print(f"Token数量: {len(tokens)}")
print(f"Token IDs: {tokens}")

# 正确方法1：整体解码
print("\n✅ 整体解码结果:")
print(encoder.decode(tokens))

# 正确方法2：逐个token解码并显示
print("\n🔍 逐个token查看:")
for i, token_id in enumerate(tokens):
    # 将单个token解码
    token_str = encoder.decode([token_id])
    
    # 如果解码结果是无效字符，显示为字节形式
    if token_str == '�' or len(token_str) == 0:
        # 获取token对应的字节
        token_bytes = encoder.decode_single_token_bytes(token_id)
        print(f"Token {i:2d} (ID: {token_id:6d}): 字节: {token_bytes} 字符: {repr(token_str)}")
    else:
        print(f"Token {i:2d} (ID: {token_id:6d}): 字符: {repr(token_str)}")

print("\n📊 总计:")
print(f"文本: {text}")
print(f"字符数: {len(text)}")
print(f"Token数: {len(tokens)}")
print(f"压缩比: {len(tokens)/len(text):.2f} tokens/字符")
```

```python
class DocumentProcessor_Correct:
    """使用完全正确模块的文档处理器"""
    
    def __init__(self, chunk_size=500, chunk_overlap=50):
        # ✅ 使用正确的 langchain_text_splitters
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            length_function=len,
            separators=["\n\n", "\n", "。", "！", "？", "；", "，", " ", ""],
            add_start_index=True,
            is_separator_regex=False
        )
    
    def load_documents(self, file_paths: List[str]) -> List[Document]:
        """加载文档"""
        all_docs = []
        
        for file_path in file_paths:
            path = Path(file_path)
            if not path.exists():
                print(f"⚠️ 文件不存在: {file_path}")
                continue
            
            try:
                if path.suffix.lower() in ['.txt', '.md']:
                    loader = TextLoader(str(path), encoding='utf-8')
                elif path.suffix.lower() == '.pdf':
                    loader = PyPDFLoader(str(path))
                else:
                    print(f"❌ 不支持的文件格式: {path.suffix}")
                    continue
                
                docs = loader.load()
                
                for i, doc in enumerate(docs):
                    doc.metadata.update({
                        "source": str(path),
                        "filename": path.name,
                        "file_type": path.suffix.lower(),
                        "doc_id": f"{path.stem}_{i}",
                        "timestamp": datetime.now().isoformat()
                    })
                
                all_docs.extend(docs)
                print(f"✅ 已加载: {path.name} ({len(docs)}个文档)")
                
            except Exception as e:
                print(f"❌ 加载失败 {path.name}: {e}")
        
        return all_docs
    
    def split_documents(self, documents: List[Document]) -> List[Document]:
        """分割文档"""
        if not documents:
            return []
        
        chunks = self.text_splitter.split_documents(documents)
        
        for i, chunk in enumerate(chunks):
            chunk.metadata.update({
                "chunk_index": i,
                "total_chunks": len(chunks),
                "chunk_hash": hashlib.md5(chunk.page_content.encode()).hexdigest()[:8]
            })
        
        print(f"📊 文档分割完成: {len(documents)}个文档 → {len(chunks)}个chunks")
        return chunks
    
    def create_sample_data(self) -> List[Document]:
        """创建示例数据"""
        sample_data = [
            {
                "content": """LangChain 1.2.x 是最新的LLM应用框架。
                注意：在1.2.x版本中，langchain.retrievers已被完全移除。
                检索器功能现在主要通过langchain_classic.retrievers导入。""",
                "metadata": {"source": "langchain_info.txt", "topic": "版本变更"}
            },
            {
                "content": """RAG系统包括文档加载、分割、向量化、检索和生成。
                在LangChain 1.2.x中，使用langchain_text_splitters进行文本分割。
                使用langchain_community.document_loaders加载文档。""",
                "metadata": {"source": "rag_system.txt", "topic": "RAG架构"}
            },
            {
                "content": """高级RAG技术包括查询扩展、重排序、混合检索。
                这些功能在1.2.x中通过langchain_classic.retrievers提供。
                包括ContextualCompressionRetriever、EnsembleRetriever等。""",
                "metadata": {"source": "advanced_rag.txt", "topic": "高级技术"}
            }
        ]
        
        documents = []
        for item in sample_data:
            doc = Document(
                page_content=item["content"],
                metadata=item["metadata"]
            )
            documents.append(doc)
        
        return documents
```

```python
class NativeRAG_Correct:
    """Native RAG - 使用完全正确的模块"""
    
    def __init__(self, persist_directory="./chroma_db_native_correct"):
        # 检查依赖
        if not HAS_CHROMA:
            raise ImportError("需要安装 langchain-chroma")
        if not HAS_OPENAI:
            raise ImportError("需要安装 langchain-openai")
        
        # 配置
        self.api_key = os.getenv("DASHSCOPE_API_KEY", "")
        self.base_url = "https://dashscope.aliyuncs.com/compatible-mode/v1"
        self.persist_directory = persist_directory
        
        # 初始化模型
        self.embeddings = self._create_embeddings()
        self.llm = self._create_llm()
        
        # 组件
        self.vectorstore = None
        self.retriever = None
        self.rag_chain = None
        
    def _create_embeddings(self):
        """创建嵌入模型"""
        return OpenAIEmbeddings(
            openai_api_key=self.api_key,
            openai_api_base=self.base_url,
            model="text-embedding-v2"
        )
    
    def _create_llm(self):
        """创建LLM"""
        return ChatOpenAI(
            openai_api_key=self.api_key,
            openai_api_base=self.base_url,
            model_name="qwen-plus",
            temperature=0.1,
            max_tokens=1000
        )
    
    def build_knowledge_base(self, documents: List[Document]):
        """构建知识库"""
        print("🔨 构建Native RAG知识库 (使用正确模块)...")
        
        # 创建向量存储
        self.vectorstore = Chroma.from_documents(
            documents=documents,
            embedding=self.embeddings,
            persist_directory=self.persist_directory
        )
        
        # 创建基础检索器
        self.retriever = self.vectorstore.as_retriever(
            search_type="similarity",
            search_kwargs={"k": 3}
        )
        
        # 构建RAG链
        self._build_rag_chain()
        
        print(f"✅ Native RAG构建完成: {len(documents)}个文档块")
        return self
    
    def _build_rag_chain(self):
        """构建RAG链"""
        template = """基于以下上下文信息回答问题：

上下文:
{context}

问题: 
{question}

如果上下文没有相关信息，请说明你不知道。

用中文回答:"""
        
        prompt = ChatPromptTemplate.from_template(template)
        
        def format_docs(docs: List[Document]) -> str:
            return "\n\n".join([doc.page_content for doc in docs])
        
        # 构建链
        self.rag_chain = (
            {
                "context": self.retriever | format_docs,
                "question": RunnablePassthrough()
            }
            | prompt
            | self.llm
            | StrOutputParser()
        )
    
    async def aquery(self, question: str) -> Dict[str, Any]:
        """异步查询"""
        if not self.rag_chain:
            raise ValueError("请先构建知识库")
        
        print(f"🔍 查询: {question}")
        
        try:
            start = datetime.now()
            answer = await self.rag_chain.ainvoke(question)
            time_used = (datetime.now() - start).total_seconds()
            
            # 获取源文档
            docs = self.retriever.invoke(question)
            
            return {
                "answer": answer,
                "time": time_used,
                "sources": [
                    {
                        "content": doc.page_content,
                        "metadata": doc.metadata
                    }
                    for doc in docs
                ],
                "source_count": len(docs)
            }
        except Exception as e:
            return {"error": str(e), "answer": f"查询失败: {e}"}
    
    def query(self, question: str) -> Dict[str, Any]:
        """同步查询"""
        return asyncio.run(self.aquery(question))
```

```python
class AdvancedRAG_Correct:
    """Advanced RAG - 使用 langchain_classic.retrievers"""
    
    def __init__(self, persist_directory="./chroma_db_advanced_correct"):
        # 检查依赖
        if not HAS_CHROMA:
            raise ImportError("需要安装 langchain-chroma")
        if not HAS_OPENAI:
            raise ImportError("需要安装 langchain-openai")
        if not HAS_CLASSIC:
            raise ImportError("需要安装 langchain-classic")
        
        # 配置
        self.api_key = os.getenv("DASHSCOPE_API_KEY", "")
        self.base_url = "https://dashscope.aliyuncs.com/compatible-mode/v1"
        self.persist_directory = persist_directory
        
        # 初始化模型
        self.embeddings = self._create_embeddings()
        self.llm = self._create_llm()
        
        # 组件
        self.vectorstore = None
        self.retriever = None
        self.rag_chain = None
        
    def _create_embeddings(self):
        """创建嵌入模型"""
        return OpenAIEmbeddings(
            openai_api_key=self.api_key,
            openai_api_base=self.base_url,
            model="text-embedding-v2"
        )
    
    def _create_llm(self):
        """创建LLM"""
        return ChatOpenAI(
            openai_api_key=self.api_key,
            openai_api_base=self.base_url,
            model_name="qwen-plus",
            temperature=0.1,
            max_tokens=1000
        )
    
    def build_knowledge_base(self, documents: List[Document]):
        """构建高级知识库"""
        print("🔨 构建Advanced RAG知识库 (使用langchain_classic)...")
        
        # 1. 创建向量存储
        self.vectorstore = Chroma.from_documents(
            documents=documents,
            embedding=self.embeddings,
            persist_directory=self.persist_directory
        )
        
        # 2. 创建基础检索器
        vector_retriever = self.vectorstore.as_retriever(
            search_type="similarity",
            search_kwargs={"k": 10}
        )
        
        # 3. ✅ 使用 langchain_classic.retrievers 创建高级检索器
        try:
            # 创建BM25检索器
            bm25_retriever = BM25Retriever.from_documents(documents)
            bm25_retriever.k = 5
            
            # 创建混合检索器
            ensemble_retriever = EnsembleRetriever(
                retrievers=[vector_retriever, bm25_retriever],
                weights=[0.7, 0.3]
            )
            
            # 创建上下文压缩检索器
            compressor = LLMChainExtractor.from_llm(self.llm)
            
            self.retriever = ContextualCompressionRetriever(
                base_compressor=compressor,
                base_retriever=ensemble_retriever
            )
            
            print("✅ 使用 langchain_classic 检索器: ContextualCompressionRetriever + EnsembleRetriever")
            
        except Exception as e:
            print(f"⚠️ 创建高级检索器失败: {e}")
            print("使用基础向量检索器")
            self.retriever = vector_retriever
        
        # 4. 构建RAG链
        self._build_rag_chain()
        
        print("✅ Advanced RAG构建完成")
        return self
    
    def _build_rag_chain(self):
        """构建RAG链"""
        template = """你是一个AI助手，基于检索到的信息回答问题。

检索到的信息:
{context}

用户问题: {question}

要求:
1. 基于检索到的信息回答
2. 如果信息不足，明确说明
3. 用中文回答
4. 结构清晰

回答:"""
        
        prompt = ChatPromptTemplate.from_template(template)
        
        def format_docs(docs: List[Document]) -> str:
            result = []
            for i, doc in enumerate(docs):
                result.append(f"[信息{i+1}]\n{doc.page_content}\n")
            return "\n".join(result)
        
        # 查询重写函数
        async def rewrite_query(query: str) -> str:
            """优化查询"""
            rewrite_prompt = ChatPromptTemplate.from_template(
                "优化这个查询以便更好地检索信息: {query}\n优化后:"
            )
            chain = rewrite_prompt | self.llm | StrOutputParser()
            
            try:
                return await chain.ainvoke({"query": query})
            except:
                return query
        
        # 构建完整pipeline
        async def full_pipeline(question: str) -> str:
            """完整的RAG pipeline"""
            # 1. 查询优化
            optimized = await rewrite_query(question)
            if optimized != question:
                print(f"🔄 查询优化: {question} -> {optimized}")
            
            # 2. 检索文档
            docs = self.retriever.invoke(optimized)
            
            # 3. 格式化上下文
            context = format_docs(docs)
            
            # 4. 生成回答
            chain = prompt | self.llm | StrOutputParser()
            answer = await chain.ainvoke({"context": context, "question": question})
            
            return answer
        
        self.rag_pipeline = full_pipeline
    
    async def aquery(self, question: str) -> Dict[str, Any]:
        """高级查询"""
        if not self.rag_pipeline:
            raise ValueError("请先构建知识库")
        
        print(f"🔍 Advanced RAG查询: {question}")
        
        try:
            start = datetime.now()
            answer = await self.rag_pipeline(question)
            time_used = (datetime.now() - start).total_seconds()
            
            # 获取源文档
            docs = self.retriever.invoke(question)
            
            return {
                "answer": answer,
                "time": time_used,
                "source_count": len(docs),
                "sources": [
                    {
                        "content": doc.page_content,
                        "metadata": doc.metadata
                    }
                    for doc in docs
                ]
            }
        except Exception as e:
            return {"error": str(e), "answer": f"查询失败: {e}"}
    
    def query(self, question: str) -> Dict[str, Any]:
        """同步查询"""
        return asyncio.run(self.aquery(question))
```

```python
class RAGWebInterface_Correct:
    """Web界面 - 对比两种RAG"""
    
    def __init__(self, native_rag: NativeRAG_Correct, advanced_rag: AdvancedRAG_Correct):
        self.native_rag = native_rag
        self.advanced_rag = advanced_rag
        self.history = []
    
    def create_interface(self):
        """创建Gradio界面"""
        
        async def process_query(question: str):
            """处理查询"""
            results = {}
            
            # Native RAG
            try:
                native_result = await self.native_rag.aquery(question)
                results["native"] = native_result
            except Exception as e:
                results["native"] = {"error": str(e)}
            
            # Advanced RAG
            try:
                advanced_result = await self.advanced_rag.aquery(question)
                results["advanced"] = advanced_result
            except Exception as e:
                results["advanced"] = {"error": str(e)}
            
            # 记录历史
            self.history.append({
                "time": datetime.now().isoformat(),
                "question": question,
                "native_time": results.get("native", {}).get("time", 0),
                "advanced_time": results.get("advanced", {}).get("time", 0)
            })
            
            # 生成对比
            comparison = self._create_comparison(results)
            details = self._create_details(results)
            
            return comparison, details
        
        def _create_comparison(results):
            """创建对比表格"""
            data = []
            
            for name, result in [("Native RAG", results.get("native")), 
                               ("Advanced RAG", results.get("advanced"))]:
                if not result or "error" in result:
                    data.append([name, f"错误: {result.get('error', '未知错误')}", "N/A", "N/A"])
                else:
                    data.append([
                        name,
                        result.get("answer", "")[:100] + ("..." if len(result.get("answer", "")) > 100 else ""),
                        f"{result.get('time', 0):.2f}s",
                        result.get("source_count", 0)
                    ])
            
            return pd.DataFrame(data, columns=["系统", "回答摘要", "时间", "来源数"])
        
        def _create_details(results):
            """创建详细结果"""
            html = "<div style='font-family: Arial, sans-serif;'>"
            
            for name, result in [("Native RAG", results.get("native")), 
                               ("Advanced RAG", results.get("advanced"))]:
                
                html += f"<h3>{name}</h3>"
                
                if not result or "error" in result:
                    html += f"<p style='color: red;'>错误: {result.get('error', '未知错误')}</p>"
                else:
                    html += f"<p><strong>回答:</strong> {result.get('answer', '')}</p>"
                    html += f"<p><strong>用时:</strong> {result.get('time', 0):.2f}秒</p>"
                    
                    if result.get("sources"):
                        html += "<h4>相关来源:</h4>"
                        for i, source in enumerate(result["sources"][:3]):
                            html += f"<div style='border-left: 3px solid #4CAF50; padding-left: 10px; margin: 10px 0;'>"
                            html += f"<p><strong>来源{i+1}:</strong> {source.get('content', '')[:200]}...</p>"
                            html += f"<p><small>元数据: {json.dumps(source.get('metadata', {}), ensure_ascii=False)}</small></p>"
                            html += "</div>"
                
                html += "<hr>"
            
            html += "</div>"
            return html
        
        # 创建界面
        with gr.Blocks(title="RAG系统对比 (正确模块)", theme=gr.themes.Soft()) as demo:
            gr.Markdown("# 🔍 Native RAG vs Advanced RAG 对比")
            gr.Markdown("使用完全正确的LangChain 1.2.x模块结构")
            
            with gr.Row():
                with gr.Column(scale=2):
                    question_input = gr.Textbox(
                        label="输入问题",
                        placeholder="例如：LangChain 1.2.x有什么变化？",
                        lines=3
                    )
                    submit_btn = gr.Button("🚀 查询", variant="primary")
                
                with gr.Column(scale=1):
                    status = gr.Textbox(label="系统状态", value="就绪")
            
            with gr.Row():
                with gr.Column():
                    gr.Markdown("### 📊 对比结果")
                    result_table = gr.Dataframe(
                        headers=["系统", "回答摘要", "时间", "来源数"],
                        interactive=False
                    )
                
                with gr.Column():
                    gr.Markdown("### 📋 详细结果")
                    details_output = gr.HTML()
            
            # 历史记录
            with gr.Tab("📜 查询历史"):
                history_df = gr.Dataframe(
                    headers=["时间", "问题", "Native时间", "Advanced时间"],
                    interactive=False
                )
            
            # 绑定事件
            submit_btn.click(
                fn=process_query,
                inputs=[question_input],
                outputs=[result_table, details_output]
            )
            
            # 更新历史
            def update_history():
                if self.history:
                    data = []
                    for item in self.history[-10:]:
                        data.append([
                            item["time"][11:19],
                            item["question"][:30] + ("..." if len(item["question"]) > 30 else ""),
                            f"{item['native_time']:.2f}s",
                            f"{item['advanced_time']:.2f}s"
                        ])
                    return pd.DataFrame(data)
                return pd.DataFrame()
            
            demo.load(fn=update_history, outputs=[history_df], every=5)
        
        return demo
```
