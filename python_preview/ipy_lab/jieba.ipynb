```python
# 将匹配的数字乘以 2
def doubled(matched):
    value = int(matched.group('value'))
    return str(value * 2)
 
s = 'A23G4HFD567'
print(re.sub('(?P<value>\d+)', doubled, s))
```

```python
# 定义处理函数（确保函数名未被覆盖）
def double_number(matched):
    # 获取命名分组的数字并转整数
    value = int(matched.group('value'))
    # 乘以2后转回字符串返回
    return str(value * 2)

# 测试字符串
s = 'A23G4HFD567'
# 关键修复：1. 正则用原始字符串r''  2. 传入正确的函数名
result = re.sub(r'(?P<value>\d+)', double_number, s)
# 输出结果
print(result)  # 正确输出：A46G8HFD1134
```

```python
%reset -f
import re
# 2. 定义处理函数（名称清晰，避免冲突）
def multiply_number_by_two(matched):
    """将匹配到的数字乘以2并返回字符串"""
    # 获取命名分组的数字
    num_str = matched.group('value')
    # 转换为整数并计算
    num = int(num_str)
    doubled_num = num * 2
    # 转回字符串返回
    return str(doubled_num)

# 3. 测试字符串
test_string = 'A23G4HFD567'

# 4. 正则替换（使用原始字符串避免转义警告）
# 关键：r'' 表示原始字符串，\d 不会被解析为转义序列
result = re.sub(r'(?P<value>\d+)', multiply_number_by_two, test_string)

# 5. 输出结果
print("原始字符串：", test_string)
print("处理后字符串：", result)
```

```python
import jieba

text_zh = "自然语言处理是人工智能的重要分支，用Python学习NLP非常方便"
# 精准模式（默认，适合常规分词）
tokens_jieba = jieba.lcut(text_zh)
print("jieba精准分词：", tokens_jieba)

# 全模式（列出所有可能的分词结果）
tokens_jieba_all = jieba.lcut(text_zh, cut_all=True)
print("jieba全模式分词：", tokens_jieba_all)

# 搜索引擎模式（适合搜索引擎分词）
tokens_jieba_search = jieba.lcut_for_search(text_zh)
print("jieba搜索引擎模式：", tokens_jieba_search)
```
